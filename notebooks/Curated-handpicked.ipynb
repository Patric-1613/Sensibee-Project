```python
import os
import shutil
from tqdm import tqdm

# 🔍 Input and Output Paths
source_root = "/content/drive/MyDrive/hand-picked-data"
output_root = "/content/drive/MyDrive/hand-picked-data-copied-clean"

# ✅ Class folders to process
class_names = ["Apoidea", "Arachnida", "Brachycera", "Coleoptera", "Formicidae", "Nematocera", "Syraphidae"]

# 📝 To collect missing labels
missing_labels = {}

print("🔁 Copying valid image-label pairs...\n")

for cls in class_names:
    src_img_dir = os.path.join(source_root, cls, "images")
    src_lbl_dir = os.path.join(source_root, cls, "labels")

    dst_img_dir = os.path.join(output_root, cls, "images")
    dst_lbl_dir = os.path.join(output_root, cls, "labels")
    os.makedirs(dst_img_dir, exist_ok=True)
    os.makedirs(dst_lbl_dir, exist_ok=True)

    missing_labels[cls] = []

    for fname in tqdm(os.listdir(src_img_dir), desc=f"Processing {cls}"):
        if not fname.endswith(".jpg"):
            continue

        base_name = os.path.splitext(fname)[0]
        img_path = os.path.join(src_img_dir, fname)
        lbl_path = os.path.join(src_lbl_dir, base_name + ".txt")

        if os.path.exists(lbl_path):
            shutil.copy(img_path, os.path.join(dst_img_dir, fname))
            shutil.copy(lbl_path, os.path.join(dst_lbl_dir, base_name + ".txt"))
        else:
            missing_labels[cls].append(fname)

# ✅ Summary of missing labels
print("\n🧾 Missing label summary:")
for cls in class_names:
    count = len(missing_labels[cls])
    print(f"• {cls}: {count} image(s) without label")
    if count > 0:
        print(f"  ⤷ Missing: {missing_labels[cls][:5]}{'...' if count > 5 else ''}")
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split
from tqdm import tqdm

# ✅ Input and Output Paths
source_root = "/content/drive/MyDrive/hand-picked-data-copied-clean"
output_root = "/content/drive/MyDrive/hand-picked-data-final-split"
train_img_dir = os.path.join(output_root, "train/images")
train_lbl_dir = os.path.join(output_root, "train/labels")
val_img_dir = os.path.join(output_root, "val/images")
val_lbl_dir = os.path.join(output_root, "val/labels")

# ✅ Make Output Directories
for d in [train_img_dir, train_lbl_dir, val_img_dir, val_lbl_dir]:
    os.makedirs(d, exist_ok=True)

# ✅ Helper to check pairs
def get_valid_image_label_pairs(class_folder):
    img_dir = os.path.join(class_folder, "images")
    lbl_dir = os.path.join(class_folder, "labels")
    pairs = []
    missing_labels = []
    if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
        return pairs, missing_labels
    for img_name in os.listdir(img_dir):
        if img_name.endswith(".jpg"):
            label_name = img_name.replace(".jpg", ".txt")
            if os.path.exists(os.path.join(lbl_dir, label_name)):
                pairs.append((os.path.join(img_dir, img_name), os.path.join(lbl_dir, label_name)))
            else:
                missing_labels.append(img_name)
    return pairs, missing_labels

# ✅ Filter only valid class folders
all_classes = [
    d for d in os.listdir(source_root)
    if os.path.isdir(os.path.join(source_root, d)) and "final-split" not in d
]

# ✅ Process each class
total_pairs, total_missing = 0, 0

for cls in tqdm(all_classes, desc="📂 Processing Classes"):
    cls_path = os.path.join(source_root, cls)
    pairs, missing = get_valid_image_label_pairs(cls_path)

    total_pairs += len(pairs)
    total_missing += len(missing)

    # Split 80:20
    train_pairs, val_pairs = train_test_split(pairs, test_size=0.2, random_state=42)

    # Copy to train
    for img_p, lbl_p in train_pairs:
        shutil.copy(img_p, os.path.join(train_img_dir, os.path.basename(img_p)))
        shutil.copy(lbl_p, os.path.join(train_lbl_dir, os.path.basename(lbl_p)))

    # Copy to val
    for img_p, lbl_p in val_pairs:
        shutil.copy(img_p, os.path.join(val_img_dir, os.path.basename(img_p)))
        shutil.copy(lbl_p, os.path.join(val_lbl_dir, os.path.basename(lbl_p)))

print(f"\n✅ Dataset split complete!")
print(f"✔️ Total images with labels used: {total_pairs}")
print(f"⚠️ Total images skipped due to missing labels: {total_missing}")
print(f"📁 Train set size: {len(os.listdir(train_img_dir))} images")
print(f"📁 Val set size:   {len(os.listdir(val_img_dir))} images")
```

```python
!pip install ultralytics
import os
import cv2
import numpy as np
from tqdm import tqdm
from ultralytics import YOLO
from sklearn.metrics import precision_score, recall_score, f1_score

# ✅ Paths
yaml_path = "/content/drive/MyDrive/hand-picked-data-final-split/config.yaml"
model_path = "models/yolo11n.pt"
output_dir = "/content/drive/MyDrive/YOLOv11_Results/reindexed_with_background_1024_curated"

# ✅ Train YOLOv11n model (no resume)
model = YOLO(model_path)
model.train(
    data=yaml_path,
    epochs=400,
    imgsz=1000,
    patience=30,
    save=True,
    project=output_dir,
    name="train_yolo11n_reindexed_with_background_1000_curated"
)

# ✅ Validation and Evaluation
val_images_path = "/content/drive/MyDrive/hand-picked-data-final-split/val/images"
val_labels_path = "/content/drive/MyDrive/hand-picked-data-final-split/val/labels"

fp_dir = os.path.join(output_dir, "false_positives")
fn_dir = os.path.join(output_dir, "false_negatives")
mc_dir = os.path.join(output_dir, "misclassified")
os.makedirs(fp_dir, exist_ok=True)
os.makedirs(fn_dir, exist_ok=True)
os.makedirs(mc_dir, exist_ok=True)

class_names = {
    0: "Apoidea", 1: "Arachnida", 2: "Brachycera", 3: "Coleoptera",
    4: "Formicidae", 5: "Nematocera", 6: "Syraphidae"
}

def compute_iou(box1, box2):
    xA, yA = max(box1[0], box2[0]), max(box1[1], box2[1])
    xB, yB = min(box1[2], box2[2]), min(box1[3], box2[3])
    inter = max(0, xB - xA) * max(0, yB - yA)
    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])
    return inter / (area1 + area2 - inter + 1e-6)

print("🔍 Evaluating on validation set...")
y_true, y_pred = [], []

for img_name in tqdm(os.listdir(val_images_path)):
    if not img_name.endswith(".jpg"):
        continue

    img_path = os.path.join(val_images_path, img_name)
    label_path = os.path.join(val_labels_path, img_name.replace(".jpg", ".txt"))
    image = cv2.imread(img_path)
    height, width = image.shape[:2]

    results = model(img_path, conf=0.25, iou=0.7)[0]
    pred_boxes = results.boxes.xyxy.cpu().numpy()
    pred_classes = results.boxes.cls.cpu().numpy()

    gt_boxes, gt_classes = [], []
    if os.path.exists(label_path):
        with open(label_path) as f:
            for line in f:
                cls, x, y, w, h = map(float, line.strip().split())
                x1 = int((x - w/2) * width)
                y1 = int((y - h/2) * height)
                x2 = int((x + w/2) * width)
                y2 = int((y + h/2) * height)
                gt_boxes.append([x1, y1, x2, y2])
                gt_classes.append(int(cls))

    matched_pred = set()
    matched_gt = set()

    for i, pbox in enumerate(pred_boxes):
        px1, py1, px2, py2 = map(int, pbox[:4])
        pred_cls = int(pred_classes[i])
        for j, gtbox in enumerate(gt_boxes):
            iou = compute_iou(pbox[:4], gtbox)
            if iou > 0.3:
                matched_pred.add(i)
                matched_gt.add(j)
                if pred_cls != gt_classes[j]:
                    img_copy = image.copy()
                    gx1, gy1, gx2, gy2 = gtbox
                    cv2.rectangle(img_copy, (gx1, gy1), (gx2, gy2), (0, 255, 0), 2)
                    cv2.rectangle(img_copy, (px1, py1), (px2, py2), (0, 0, 255), 2)
                    cv2.putText(img_copy, f"Pred: {class_names.get(pred_cls)}", (px1, py1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,255), 2)
                    cv2.putText(img_copy, f"GT: {class_names.get(gt_classes[j])}", (gx1, gy1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
                    cv2.imwrite(os.path.join(mc_dir, f"mc_{img_name}"), img_copy)
                break

    for i, pbox in enumerate(pred_boxes):
        if i not in matched_pred:
            y_true.append(0)
            y_pred.append(1)
            px1, py1, px2, py2 = map(int, pbox[:4])
            pred_cls = int(pred_classes[i])
            img_copy = image.copy()
            cv2.rectangle(img_copy, (px1, py1), (px2, py2), (0, 0, 255), 2)
            cv2.putText(img_copy, f"FP: {class_names.get(pred_cls)}", (px1, py1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,255), 2)
            cv2.imwrite(os.path.join(fp_dir, f"fp_{img_name}"), img_copy)

    for j, gtbox in enumerate(gt_boxes):
        if j not in matched_gt:
            y_true.append(1)
            y_pred.append(0)
            gx1, gy1, gx2, gy2 = gtbox
            img_copy = image.copy()
            cv2.rectangle(img_copy, (gx1, gy1), (gx2, gy2), (255, 0, 0), 2)
            cv2.putText(img_copy, "FN", (gx1, gy1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)
            cv2.imwrite(os.path.join(fn_dir, f"fn_{img_name}"), img_copy)

# ✅ Summary metrics
precision = precision_score(y_true, y_pred) * 100
recall = recall_score(y_true, y_pred) * 100
f1 = f1_score(y_true, y_pred) * 100

print("\n✅ Evaluation Summary:")
print(f"Precision: {precision:.2f}%")
print(f"Recall:    {recall:.2f}%")
print(f"F1 Score:  {f1:.2f}%")
print(f"→ FP saved to: {fp_dir}")
print(f"→ FN saved to: {fn_dir}")
print(f"→ MC saved to: {mc_dir}")
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split
from tqdm import tqdm

# ✅ Source and output directories
source_root = "/content/drive/MyDrive/hand-picked-data-copied-clean"
output_root = "/content/drive/MyDrive/hand-picked-data-final-split-corrected"

# Define output structure
splits = ["train", "val"]
subdirs = ["images", "labels"]
for split in splits:
    for subdir in subdirs:
        os.makedirs(os.path.join(output_root, split, subdir), exist_ok=True)

# ✅ Process each class folder
for class_name in sorted(os.listdir(source_root)):
    class_path = os.path.join(source_root, class_name)
    if not os.path.isdir(class_path):
        continue

    img_dir = os.path.join(class_path, "images")
    lbl_dir = os.path.join(class_path, "labels")
    if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
        print(f"⚠️ Skipping {class_name}: Missing 'images/' or 'labels/'")
        continue

    # Collect valid (image, label) pairs
    all_pairs = []
    for fname in os.listdir(img_dir):
        if fname.endswith(".jpg"):
            image_path = os.path.join(img_dir, fname)
            label_path = os.path.join(lbl_dir, fname.replace(".jpg", ".txt"))
            if os.path.exists(label_path):
                all_pairs.append((image_path, label_path))

    if not all_pairs:
        print(f"⚠️ No valid pairs found in class: {class_name}")
        continue

    # 80:20 split
    train_pairs, val_pairs = train_test_split(all_pairs, test_size=0.2, random_state=42)

    # Copy files to split directories
    for split_name, pairs in zip(["train", "val"], [train_pairs, val_pairs]):
        for img_path, lbl_path in pairs:
            base_name = os.path.basename(img_path)
            shutil.copy(img_path, os.path.join(output_root, split_name, "images", base_name))
            shutil.copy(lbl_path, os.path.join(output_root, split_name, "labels", base_name.replace(".jpg", ".txt")))

    print(f"✅ {class_name}: {len(train_pairs)} train | {len(val_pairs)} val")

print("\n🎉 Dataset split complete and saved to:", output_root)
```

```python
import os
from tqdm import tqdm

# ✅ Class-to-ID mapping
class_to_id = {
    "Apoidea": 0,
    "Arachnida": 1,
    "Brachycera": 2,
    "Coleoptera": 3,
    "Formicidae": 4,
    "Nematocera": 5,
    "Syraphidae": 6
}

# ✅ Root path to your class-wise dataset
base_dir = "/content/drive/MyDrive/hand-picked-data-copied-clean"

# ✅ Track how many label files updated per class
update_counts = {}

for class_name, class_id in class_to_id.items():
    class_folder = os.path.join(base_dir, class_name, "labels")
    if not os.path.exists(class_folder):
        print(f"⚠️ Skipping {class_name} (no labels folder found)")
        continue

    updated = 0
    for file in tqdm(os.listdir(class_folder), desc=f"Rewriting {class_name}"):
        if file.endswith(".txt"):
            file_path = os.path.join(class_folder, file)
            with open(file_path, "r") as f:
                lines = f.readlines()

            # Overwrite with correct class ID
            updated_lines = []
            for line in lines:
                parts = line.strip().split()
                if len(parts) == 5:  # valid label line
                    parts[0] = str(class_id)  # update class ID
                    updated_lines.append(" ".join(parts))

            # Only write back if lines are valid
            if updated_lines:
                with open(file_path, "w") as f:
                    f.write("\n".join(updated_lines) + "\n")
                updated += 1

    update_counts[class_name] = updated

# ✅ Print summary
print("\n📊 Label Overwrite Summary:")
for cls, count in update_counts.items():
    print(f"  {cls}: {count} label files updated")
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split
from tqdm import tqdm

# ✅ Source and output directories
source_root = "/content/drive/MyDrive/hand-picked-data-copied-clean"
output_root = "/content/drive/MyDrive/hand-picked-data-final-split-corrected_again"

# Define output structure
splits = ["train", "val"]
subdirs = ["images", "labels"]
for split in splits:
    for subdir in subdirs:
        os.makedirs(os.path.join(output_root, split, subdir), exist_ok=True)

# ✅ Process each class folder
for class_name in sorted(os.listdir(source_root)):
    class_path = os.path.join(source_root, class_name)
    if not os.path.isdir(class_path):
        continue

    img_dir = os.path.join(class_path, "images")
    lbl_dir = os.path.join(class_path, "labels")
    if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
        print(f"⚠️ Skipping {class_name}: Missing 'images/' or 'labels/'")
        continue

    # Collect valid (image, label) pairs
    all_pairs = []
    for fname in os.listdir(img_dir):
        if fname.endswith(".jpg"):
            image_path = os.path.join(img_dir, fname)
            label_path = os.path.join(lbl_dir, fname.replace(".jpg", ".txt"))
            if os.path.exists(label_path):
                all_pairs.append((image_path, label_path))

    if not all_pairs:
        print(f"⚠️ No valid pairs found in class: {class_name}")
        continue

    # 80:20 split
    train_pairs, val_pairs = train_test_split(all_pairs, test_size=0.2, random_state=42)

    # Copy files to split directories
    for split_name, pairs in zip(["train", "val"], [train_pairs, val_pairs]):
        for img_path, lbl_path in pairs:
            base_name = os.path.basename(img_path)
            shutil.copy(img_path, os.path.join(output_root, split_name, "images", base_name))
            shutil.copy(lbl_path, os.path.join(output_root, split_name, "labels", base_name.replace(".jpg", ".txt")))

    print(f"✅ {class_name}: {len(train_pairs)} train | {len(val_pairs)} val")

print("\n🎉 Dataset split complete and saved to:", output_root)
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

# ✅ Paths
bg_source_dir = "/content/drive/MyDrive/no-tags-3"
dataset_dir = "/content/drive/MyDrive/hand-picked-data-final-split-corrected_again"
train_img_dir = os.path.join(dataset_dir, "train/images")
val_img_dir = os.path.join(dataset_dir, "val/images")

# ✅ Get all background image paths
bg_images = [f for f in os.listdir(bg_source_dir) if f.endswith(".jpg")]
train_bg, val_bg = train_test_split(bg_images, test_size=0.2, random_state=42)

# ✅ Count before adding
before_train = len(os.listdir(train_img_dir))
before_val = len(os.listdir(val_img_dir))

# ✅ Copy background images
for img in train_bg:
    shutil.copy(os.path.join(bg_source_dir, img), os.path.join(train_img_dir, img))

for img in val_bg:
    shutil.copy(os.path.join(bg_source_dir, img), os.path.join(val_img_dir, img))

# ✅ Count after adding
after_train = len(os.listdir(train_img_dir))
after_val = len(os.listdir(val_img_dir))

# ✅ Report
print("✅ Background-only images added.")
print(f"🔢 Background images added to train: {len(train_bg)}")
print(f"🔢 Background images added to val:   {len(val_bg)}")
print(f"📈 Total images in train before: {before_train} → after: {after_train}")
print(f"📈 Total images in val before:   {before_val} → after: {after_val}")
```

```python
# !pip install ultralytics
import os
import cv2
import numpy as np
from tqdm import tqdm
from ultralytics import YOLO
from sklearn.metrics import precision_score, recall_score, f1_score

# ✅ Paths
yaml_path = "/content/drive/MyDrive/hand-picked-data-final-split-corrected_again/config.yaml"
model_path = "models/yolo11n.pt"
output_dir = "/content/drive/MyDrive/YOLOv11_Results/reindexed_with_background_1024_curated_corrected_again"

# ✅ Train YOLOv11n model (no resume)
model = YOLO(model_path)
model.train(
    data=yaml_path,
    epochs=400,
    imgsz=1000,
    patience=30,
    save=True,
    project=output_dir,
    name="train_yolo11n_reindexed_with_background_1000_curated_corrected_again"
)

# ✅ Validation and Evaluation
val_images_path = "/content/drive/MyDrive/hand-picked-data-final-split-corrected_again/val/images"
val_labels_path = "/content/drive/MyDrive/hand-picked-data-final-split-corrected_again/val/labels"

fp_dir = os.path.join(output_dir, "false_positives")
fn_dir = os.path.join(output_dir, "false_negatives")
mc_dir = os.path.join(output_dir, "misclassified")
os.makedirs(fp_dir, exist_ok=True)
os.makedirs(fn_dir, exist_ok=True)
os.makedirs(mc_dir, exist_ok=True)

class_names = {
    0: "Apoidea", 1: "Arachnida", 2: "Brachycera", 3: "Coleoptera",
    4: "Formicidae", 5: "Nematocera", 6: "Syraphidae"
}

def compute_iou(box1, box2):
    xA, yA = max(box1[0], box2[0]), max(box1[1], box2[1])
    xB, yB = min(box1[2], box2[2]), min(box1[3], box2[3])
    inter = max(0, xB - xA) * max(0, yB - yA)
    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])
    return inter / (area1 + area2 - inter + 1e-6)

print("🔍 Evaluating on validation set...")
y_true, y_pred = [], []

for img_name in tqdm(os.listdir(val_images_path)):
    if not img_name.endswith(".jpg"):
        continue

    img_path = os.path.join(val_images_path, img_name)
    label_path = os.path.join(val_labels_path, img_name.replace(".jpg", ".txt"))
    image = cv2.imread(img_path)
    height, width = image.shape[:2]

    results = model(img_path, conf=0.25, iou=0.7)[0]
    pred_boxes = results.boxes.xyxy.cpu().numpy()
    pred_classes = results.boxes.cls.cpu().numpy()

    gt_boxes, gt_classes = [], []
    if os.path.exists(label_path):
        with open(label_path) as f:
            for line in f:
                cls, x, y, w, h = map(float, line.strip().split())
                x1 = int((x - w/2) * width)
                y1 = int((y - h/2) * height)
                x2 = int((x + w/2) * width)
                y2 = int((y + h/2) * height)
                gt_boxes.append([x1, y1, x2, y2])
                gt_classes.append(int(cls))

    matched_pred = set()
    matched_gt = set()

    for i, pbox in enumerate(pred_boxes):
        px1, py1, px2, py2 = map(int, pbox[:4])
        pred_cls = int(pred_classes[i])
        for j, gtbox in enumerate(gt_boxes):
            iou = compute_iou(pbox[:4], gtbox)
            if iou > 0.3:
                matched_pred.add(i)
                matched_gt.add(j)
                if pred_cls != gt_classes[j]:
                    img_copy = image.copy()
                    gx1, gy1, gx2, gy2 = gtbox
                    cv2.rectangle(img_copy, (gx1, gy1), (gx2, gy2), (0, 255, 0), 2)
                    cv2.rectangle(img_copy, (px1, py1), (px2, py2), (0, 0, 255), 2)
                    cv2.putText(img_copy, f"Pred: {class_names.get(pred_cls)}", (px1, py1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,255), 2)
                    cv2.putText(img_copy, f"GT: {class_names.get(gt_classes[j])}", (gx1, gy1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
                    cv2.imwrite(os.path.join(mc_dir, f"mc_{img_name}"), img_copy)
                break

    for i, pbox in enumerate(pred_boxes):
        if i not in matched_pred:
            y_true.append(0)
            y_pred.append(1)
            px1, py1, px2, py2 = map(int, pbox[:4])
            pred_cls = int(pred_classes[i])
            img_copy = image.copy()
            cv2.rectangle(img_copy, (px1, py1), (px2, py2), (0, 0, 255), 2)
            cv2.putText(img_copy, f"FP: {class_names.get(pred_cls)}", (px1, py1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,255), 2)
            cv2.imwrite(os.path.join(fp_dir, f"fp_{img_name}"), img_copy)

    for j, gtbox in enumerate(gt_boxes):
        if j not in matched_gt:
            y_true.append(1)
            y_pred.append(0)
            gx1, gy1, gx2, gy2 = gtbox
            img_copy = image.copy()
            cv2.rectangle(img_copy, (gx1, gy1), (gx2, gy2), (255, 0, 0), 2)
            cv2.putText(img_copy, "FN", (gx1, gy1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)
            cv2.imwrite(os.path.join(fn_dir, f"fn_{img_name}"), img_copy)

# ✅ Summary metrics
precision = precision_score(y_true, y_pred) * 100
recall = recall_score(y_true, y_pred) * 100
f1 = f1_score(y_true, y_pred) * 100

print("\n✅ Evaluation Summary:")
print(f"Precision: {precision:.2f}%")
print(f"Recall:    {recall:.2f}%")
print(f"F1 Score:  {f1:.2f}%")
print(f"→ FP saved to: {fp_dir}")
print(f"→ FN saved to: {fn_dir}")
print(f"→ MC saved to: {mc_dir}")
```
